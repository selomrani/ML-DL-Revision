### PART - B

```python
age = dataset['age'].to_numpy()
fare = dataset['fare'].to_numpy()

np.isnan(age).sum() ## 177 misiing value
print(np.mean(age)) # outputs NAN
print(np.mean(age[~np.isnan(age)])) # No error

age_mean = np.mean(age[~np.isnan(age)])
age_med = np.median(age[~np.isnan(age)])
age_std = np.std(age[~np.isnan(age)])
age_min = np.min(age[~np.isnan(age)])
age_max = np.max(age[~np.isnan(age)])
age_25_iqr = np.percentile(age[~np.isnan(age)], 25)
age_75_iqr = np.percentile(age[~np.isnan(age)], 75)

fare_mean = np.mean(fare[~np.isnan(fare)])
fare_med = np.median(fare[~np.isnan(fare)])
fare_std = np.std(fare[~np.isnan(fare)])
fare_min = np.min(fare[~np.isnan(fare)])
fare_max = np.max(fare[~np.isnan(fare)])
fare_25_iqr = np.percentile(fare[~np.isnan(fare)], 25)
fare_75_iqr = np.percentile(fare[~np.isnan(fare)], 75)

# note to self , i can see here that the mean is sensitive to outliers , while the median is not , and the std is sensitive to outliers , and the data is skewed to the right

age_clean = age[~np.isnan(age)]
fare_clean = fare[~np.isnan(fare)]

q1, q3 = np.percentile(fare_clean, [25, 75])
iqr = q3 - q1
borne_sup = q3 + 1.5 * iqr

outliers = fare_clean > borne_sup
print(iqr, borne_sup)
print(outliers.sum())   # number of outliers
```

## logarithmic transformation


```python
logo_fare = np.log1p(fare_clean)
np.mean(logo_fare)
np.median(logo_fare)
# i can see here that the data is more normally distributed after the logarithmic transformation
```

## Correlation

```python
valid = ~np.isnan(age) & ~np.isnan(fare)

r = np.corrcoef(age[valid], fare[valid])[0, 1]
print(r)

# i can see here that the correlation is close to 0 , which means that there is no linear relationship between age and fare
```
