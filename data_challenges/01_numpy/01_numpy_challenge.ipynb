## Boolean Mask

```python
mask = (matrix > 8) & (matrix % 2 == 0)
mask
print(mask)
# now we replace them with zeroes in the matrix itself
matrix[mask] = 0
matrix
```

```python
matrix = np.arange(20).reshape(5, 4)
condition = matrix % 5 == 0

indexs , column = np.where(condition)
indexs
```

## Copy and view diff

```python
view = matrix[1:4]
view[0][0] = 100
print(matrix)
matrix = np.arange(20).reshape(5, 4)
copy = matrix[1:4].copy()
copy[0,0] = 100
print(matrix)

## Note to self , copy is an indepedent element than the og element , manipulations done on the copy doesnt effect the og element  , while the viw is a view of the og element and any manipulations done on the view will effect the og element , because they share the same place in memory
```
